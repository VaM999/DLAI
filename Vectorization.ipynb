```python
import numpy as np
from codecarbon import EmissionsTracker
import time
```

```python
tracker = EmissionsTracker(output_methods=[], log_level="error")
tracker.start()

try:
    a = np.random.rand(1000000)
    b = np.random.rand(1000000)

    start_time = time.time()
    c = np.dot(a, b)
    stop_time = time.time()

    print("Vectorized dot product time duration: " + str(stop_time - start_time) + " seconds")
finally:
    emissions = tracker.stop()
print(f"Emissions: {emissions} kg CO2")
```

```python
tracker = EmissionsTracker(output_methods=[], log_level="error")
tracker.start()

try:
    a = np.random.rand(1000000)
    b = np.random.rand(1000000)

    start_time = time.time()

    c = 0
    for i in range(0, 1000000):
        c = c + a[i] * b[i]

    stop_time = time.time()
    print("Non vectorized (Iterated) dot product time duration: " + str(stop_time - start_time) + " seconds")
finally:
    emissions = tracker.stop()
print(f"Emissions: {emissions} kg CO2")
```
